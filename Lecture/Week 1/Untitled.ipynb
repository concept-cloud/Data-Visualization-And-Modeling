```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("assets/titanic_passengers.csv")

df.head(10)
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Fare"], bins=20)
plt.title("Titanic Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.show()

print("Minimum Fare:", df["Fare"].min())
print("Maximum Fare:", df["Fare"].max())
print("Average Fare:", round(df["Fare"].mean(), 2))
print("Median Fare:", round(df["Fare"].median(), 2))
```

```python
df["Survived"].value_counts().sort_index().plot(
    kind="bar"
)

plt.title("Titanic Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Passengers")
plt.show

survival_pct = df["Survived"].value_counts(normalize=True) * 100

print(survival_pct)
```

```python
sex_counts = df["Sex"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    sex_counts,
    labels=sex_counts.index,
    autopct="%1.1f%%"
)

plt.title("Passenger Sex")
plt.show()

sex_pct = df["Sex"].value_counts(normalize=True) * 100

print(sex_pct)
```
